```python
import json

def pretty_print_data(data_dict):
    """
    User-defined function to pretty print all aggregated user activity,
    network info, personal details, and religion/community fields.
    """
    border = "=" * 70
    
    print(border)
    print("                        PARSED DATA SUMMARY                     ")
    print(border)
    
    for section, details in data_dict.items():
        title = section.replace("_", " ").upper()
        print(f"\n▶ {title}:")
        
        if isinstance(details, dict):
            for sub_key, sub_val in details.items():
                formatted_key = sub_key.replace("_", " ").title()
                if isinstance(sub_val, list):
                    print(f"   • {formatted_key}:")
                    for item in sub_val:
                        print(f"       - {item}")
                else:
                    print(f"   • {formatted_key}: {sub_val}")
        elif isinstance(details, list):
            for item in details:
                print(f"   • {item}")
        else:
            print(f"   {details}")

    print(f"\n{border}")
    print("                         RAW JSON VIEW                          ")
    print(border)
    print(json.dumps(data_dict, indent=4))
    print(border)


# Aggregated data structure updated with religion and community information
all_data = {
    "device_and_network_info": {
        "ip_address": "192.168.1.12",
        "times_opened_device": "3:4",
        "account_link": "https://myaccount.google.com/",
        "live_location": "https://www.google.com/maps/place/Shri+Ram+Mandir+Temple,Jewargi+Road/@17.2992184,76.8176726,1331m/",
        "most_visited_servers": [
            "ChatGPT / OpenAI",
            "Google",
            "YouTube",
            "Facebook"
        ]
    },
    "leak_metadata": {
        "source": "HiTeckGroop.in",
        "description": "2025 Indian cellular operator leak (~1.8B records, ~300M unique users)."
    },
    "subscriber_record": {
        "full_name": "D O SHRIKRISHNA DOLE",
        "father_name": "SHWETA",
        "telephone": "918050019998",
        "address": "H No D6/4 CADA Quate B B Road BHEEMARAYANAGUDI BHEEMARAYANAGUDI Bheemarayanagudi Yadgir KARNATAKA 585287",
        "document_number": "[Aadhaar Redacted]",
        "region": "BSNL KARNATKA",
        "religion": "Hindu",
        "caste_community": "Brahmin"
    }
}

# Run the user-defined pretty print function
pretty_print_data(all_data)
```
